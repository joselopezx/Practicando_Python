#Algoritmos Metaheurísticos
Considere α = 0.1; h = 0.1 e implemente el gradiente descendente para hallar el valor de “x”
que minimice las funciones según sea el caso. Elija la cantidad de iteraciones que se adecue
al problema.


```python
import matplotlib.pyplot as plt
import numpy as np
import random
import pandas as pd

a=0.1
h=0.1
ITERACIONES = 30
```

## 1. Hallar el valor de x que minimice la función:
f(x, y) = −e^(
−(x
2+y
2
))

[x, y] = [−1,1]

```python
def f(x, y):
    return -np.exp(-(x**2 + y**2))

alpha = 0.1
h = 0.1
iteraciones = 50

x, y = np.random.uniform(-1, 1), np.random.uniform(-1, 1)

print("Iteración| x     | y     | f(x,y)")
for k in range(iteraciones):
    gx = (f(x + h, y) - f(x, y)) / h
    gy = (f(x, y + h) - f(x, y)) / h

    x = x - alpha * gx
    y = y - alpha * gy

    x = np.clip(x, -1, 1)
    y = np.clip(y, -1, 1)

    print(f"{k:3d}      | {x:.4f} | {y:.4f} | {f(x,y):.6f}")
```

## 2. Hallar el valor de x que minimice la función:
f(x) = x^2 + 2x + 1

Considerando que x = [−10, 10]

```python
x = random.randint(-10, 10)
inicio = x
valoresX = []
valoresY = []
for iteracion in range (ITERACIONES):
  valoresX.append(x)
  valoresY.append(x**2 + 2*x + 1)
  x = x - a*(2*x + 2)


x_vals = np.linspace(inicio, -1*inicio, 400)
y_vals = x_vals**2 + 2*x_vals + 1

dataFrame = pd.DataFrame(valoresY, valoresX)
print(dataFrame)

plt.figure(figsize=(8,5))
plt.plot(x_vals, y_vals, label="f(x) = x² + 2x + 1")
plt.scatter(valoresX, valoresY, color='red', label="Puntos iterativos", s=50)
plt.title(f"Valor X inicial = {inicio}\nValor aproximado {valoresX[-1]}")

plt.show()
```

##3. Hallar el valor de x que maximice la función:
f(x) = −x^2 + 4x − 3

Considerando que x = [−5, 5]

```python
x = random.randint(-5, 5)
inicio = x
valoresX = []
valoresY = []
for iteracion in range (ITERACIONES):
  valoresX.append(x)
  valoresY.append(-1*(-x**2 + 4*x - 3))
  x = x - a*(2*x - 4)

x_vals = np.linspace(-5, 5, 400)
y_vals = -1*(-x_vals**2 + 4*x_vals - 3)


plt.figure(figsize=(8,5))
plt.plot(x_vals, y_vals, label="f(x) = x**2 - 4*x + 3")
plt.scatter(valoresX, valoresY, color='red', label="Puntos iterativos", s=50)
plt.title(f"Valor X inicial = {inicio}\nValor aproximado {valoresX[-1]}")

plt.show()
```
